```python
import random
```

```python
import pandas as pd
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
lst = ['robot'] * 10
lst += ['human'] * 10
random.shuffle(lst)
```

```python
data = pd.DataFrame({'whoAmI': lst})
```

```python
unique_values = list(set(data['whoAmI']))
one_hot = pd.DataFrame()
```

```python
for value in unique_values:
    one_hot[value] = (data['whoAmI'] == value).astype(int)
```

```python
print("Исходный DataFrame:")
print(data.head(10))
print("\nOne Hot Encoding:")
print(one_hot.head(10))
```

```python
print("\nПроверка размерности:")
print(f"Количество строк в исходном DataFrame: {len(data)}")
print(f"Количество строк в One Hot: {len(one_hot)}")
```

```python
print("\nПроверка корректности One Hot Encoding:")
print(f"Сумма значений по строкам: \n{one_hot.sum(axis=1).value_counts()}")
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(one_hot.head(10), cmap='coolwarm', annot=True, fmt='d', cbar=False)
plt.title('Визуализация One Hot Encoding (первые 10 строк)')
plt.show()
```

```python
print("\nСравнение с pd.get_dummies():")
pd_dummies = pd.get_dummies(data['whoAmI'])
print("\nРезультат pd.get_dummies():")
print(pd_dummies.head(10))
```

```python
print("\nПроверка совпадения значений по столбцам:")
for column in pd_dummies.columns:
    column_equal = (pd_dummies[column] == one_hot[column]).all()
    print(f"Столбец {column}: {column_equal}")
```

```python
data_encoded = pd.get_dummies(data['whoAmI'])
```

```python
print("\nOne Hot Encoding результат:")
display(data_encoded.head())
```

```python
plt.figure(figsize=(10, 5))
sns.heatmap(data_encoded.head(10), annot=True, fmt='d', cmap='coolwarm', cbar=False)
plt.title('Визуализация One Hot Encoding (первые 10 строк)')
plt.show()
```

```python
print("\nПроверка размерности:")
print(f"Исходный DataFrame: {data.shape}")
print(f"One Hot Encoding DataFrame: {data_encoded.shape}")
```

```python
print("\nПроверка сумм по строкам (должны быть равны 1):")
print(data_encoded.sum(axis=1).value_counts())
```

```python
print("\nСтатистика по столбцам:")
print(data_encoded.describe())
```
